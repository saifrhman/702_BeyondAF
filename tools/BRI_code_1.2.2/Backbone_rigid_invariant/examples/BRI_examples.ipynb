## Extended use
These classes are intended to load minimal data for BRI computation. However, if there are any information interests you in the cif file, you can also add them into the object by customizing `extra_keys` parameter as `Dict[name_you_want, category_in_cif]`


```python
entry_extra = Entry('2hhb', extra_keys={'true_database': 'database_2', 'false_database': 'database_not_in_cif'})
entry_extra
```

```python
entry_extra.true_database
```

```python
# non-existing category gives nothing
entry_extra.false_database
```
